```python
def total_reads(counts):
    total = 0
    for value in counts:
        total = total + value
    return total
```

```python
assert total_reads([1, 2, 3, 4]) == 10, total_reads([1, 2, 3, 4])
assert total_reads([]) == 0, total_reads([])
print("A1 ok    total =", total_reads(counts))
```
